```python
df.isna().sum().sort_values(ascending=False)
```

Missing values are limited to two columns: `album_name` and `track_name`.  
`album_name` has a small number of missing values and can be filled with "Unknown" during the cleaning phase.  
`track_name` has very few missing values and these rows may be removed because the track title is important for identification and dashboard reporting.

```python
missing_percentage = (df.isna().sum() / len(df) * 100).round(2)

missing_percentage.sort_values(ascending=False)
```

The dataset has a very low percentage of missing values. Only `album_name` and `track_name` contain missing data, with missing percentages of 0.05% and 0.02% respectively. Since the percentage is extremely low, these issues can be handled easily during the cleaning phase.

```python
df["track_id"].duplicated().sum()
```

No duplicated rows were found in the dataset. The `track_id` column also contains no duplicated values, meaning that each track has a unique identifier. This makes `track_id` suitable as a primary key for future SQL database modeling.

```python
df.nunique().sort_values(ascending=False)
```

The uniqueness check confirms that `track_id` is unique for all 85,000 rows, making it suitable as a primary key. Categorical variables such as `genre`, `country`, and `label` have a limited number of unique values, which makes them suitable for grouped analysis and dashboard visualizations. High-cardinality fields such as `track_name`, `artist_name`, and `album_name` should be used carefully in aggregations.

```python
df["genre"].value_counts()
```

The dataset contains 12 music genres with a very balanced distribution. Each genre contains approximately 6,900 to 7,200 tracks. This balance allows reliable comparisons across genres, although it may not fully reflect the real-world distribution of music available on Spotify.

```python
df["country"].value_counts()
```

The dataset includes 10 countries with a balanced number of tracks per country, ranging from approximately 8,300 to 8,600 tracks. This makes country-level comparisons possible, although the distribution may not fully reflect real-world Spotify market proportions.

```python
df["label"].value_counts()
```

The dataset includes 8 label categories with a balanced distribution. This enables comparisons between labels, including Independent tracks versus major labels. However, the uniform distribution may not fully represent real-world market shares.

```python
quality_report = pd.DataFrame({
    "column": df.columns,
    "dtype": df.dtypes.astype(str).values,
    "missing_values": df.isna().sum().values,
    "missing_percentage": (df.isna().sum().values / len(df) * 100).round(2),
    "unique_values": df.nunique().values
})

quality_report
```

```python
quality_report.to_csv(
    quality_reports_path + "/spotify_data_quality_report.csv",
    index=False
)
```
